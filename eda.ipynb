```python
import pandas as pd

df2 = pd.read_csv("data/raw/smhi-opendata_19_71380_200603_201602.csv", sep=";", skiprows=11)
df1 = pd.read_csv("data/raw/smhi-opendata_19_71380_201602_202601.csv", sep=";", skiprows=11)
```

### Concat all dataframes

```python
all_df = pd.concat([df1, df2], ignore_index=True)

all_df.head()
```

```python
all_df.info()
```

### Change names on columns

```python
data = { 
    'Representativt dygn': 'date',
    'Lufttemperatur': 'temp_min',
    'Lufttemperatur.1': 'temp_max',
    'Kvalitet': 'quality_min',
    'Kvalitet.1': 'quality_max'
}

all_df = all_df.rename(columns=data)

all_df
```

###  Take away unnecssary columns

```python
all_df = all_df.drop(columns=['Unnamed: 7', 'Tidsutsnitt:'])

all_df.head()
```

### Convert date

```python
all_df['date'] = pd.to_datetime(all_df['date'])
```

```python
all_df.info()
```

```python
df = all_df.set_index('date').sort_index()

df.head()
```

### Deal with qualityflags & make it binary

```python
all_df['quality_flag'] = (
    (all_df['quality_min'] != 'G') |
    (all_df['quality_max'] != 'G')
).astype(int)

all_df.sample(5)
```

```python
duplicates = all_df[all_df.duplicated()]

duplicates
```

```python
duplicates = all_df.drop_duplicates()

all_df = duplicates

all_df
```

```python
all_df = all_df.drop(columns=["Från Datum Tid (UTC)", "Till Datum Tid (UTC)"])

```
